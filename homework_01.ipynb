```python
X = (
    df[df["origin"] == "Asia"]
    [["vehicle_weight", "model_year"]]
    .head(7)
    .to_numpy()
)
```

```python
X
```

```python
X.shape
```

```python
X.T
```

```python
X.T.shape
```

```python
XTX = X.T @ X
XTX
```

```python
XTX.shape
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
w = XTX_inv @ X.T @ y
w
```
